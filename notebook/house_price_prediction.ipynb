```python
import pandas as pd

df = pd.read_csv("../data/raw/California_Houses.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
print("Data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df.describe().T
```

```python
target_column = "Median_House_Value"

X = df.drop(columns=[target_column])
y = df[target_column]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget column:")
print(y.name)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.hist(y, bins=50)
plt.xlabel("Median House Value")
plt.ylabel("Number of Houses")
plt.title("Distribution of Median House Values")
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.boxplot(y, vert=False)
plt.xlabel("Median House Value")
plt.title("Box Plot of Median House Values")
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
print("Feature ranges:")

feature_ranges = pd.DataFrame({
    "Minimum": X_train.min(),
    "Maximum": X_train.max(),
    "Mean": X_train.mean(),
    "Standard Deviation": X_train.std()
})

display(feature_ranges)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

preprocessing_pipeline = Pipeline([
    ("scaler", StandardScaler())
])

X_train_scaled = preprocessing_pipeline.fit_transform(X_train)
X_test_scaled = preprocessing_pipeline.transform(X_test)

print("Preprocessing completed successfully!")
print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(X_train_scaled, y_train)

print("Linear Regression model trained successfully!")
```

```python
y_pred_linear = linear_model.predict(X_test_scaled)

print("Predictions generated successfully!")
print("Number of predictions:", len(y_pred_linear))

print("\nFirst 5 predictions:")
print(y_pred_linear[:5])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae_linear = mean_absolute_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mean_squared_error(y_test, y_pred_linear))
r2_linear = r2_score(y_test, y_pred_linear)

print("Linear Regression Performance")
print("-" * 35)
print(f"MAE  : ${mae_linear:,.2f}")
print(f"RMSE : ${rmse_linear:,.2f}")
print(f"R²   : {r2_linear:.4f}")
```

```python
residuals_linear = y_test - y_pred_linear

plt.figure(figsize=(10, 5))
plt.scatter(y_pred_linear, residuals_linear, alpha=0.4)
plt.axhline(y=0, linestyle="--")
plt.xlabel("Predicted House Value")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot — Linear Regression")
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

random_forest_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
y_pred_rf = random_forest_model.predict(X_test)

mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
r2_rf = r2_score(y_test, y_pred_rf)

print("Random Forest Performance")
print("-" * 35)
print(f"MAE  : ${mae_rf:,.2f}")
print(f"RMSE : ${rmse_rf:,.2f}")
print(f"R²   : {r2_rf:.4f}")
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae_linear, mae_rf],
    "RMSE": [rmse_linear, rmse_rf],
    "R2": [r2_linear, r2_rf]
})

display(comparison)
```

```python
sample_predictions = pd.DataFrame({
    "Actual House Value": y_test.iloc[:5].values,
    "Predicted House Value": y_pred_rf[:5]
})

sample_predictions["Absolute Error"] = (
    sample_predictions["Actual House Value"]
    - sample_predictions["Predicted House Value"]
).abs()

display(sample_predictions)
```

```python
residuals_rf = y_test - y_pred_rf

plt.figure(figsize=(10, 5))
plt.scatter(y_pred_rf, residuals_rf, alpha=0.4)
plt.axhline(y=0, linestyle="--")
plt.xlabel("Predicted House Value")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot — Random Forest")
plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": random_forest_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(feature_importance)
```

```python
plt.figure(figsize=(10, 6))

plt.barh(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Random Forest Feature Importance")
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
# Save model evaluation metrics

metrics = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae_linear, mae_rf],
    "RMSE": [rmse_linear, rmse_rf],
    "R2": [r2_linear, r2_rf]
})

metrics.to_csv("../outputs/metrics.csv", index=False)

print("Metrics saved successfully!")
print("File: outputs/metrics.csv")

display(metrics)
```

```python
# Save 5 sample predictions

sample_predictions = pd.DataFrame({
    "Actual House Value": y_test.iloc[:5].values,
    "Predicted House Value": y_pred_rf[:5]
})

sample_predictions["Absolute Error"] = (
    sample_predictions["Actual House Value"]
    - sample_predictions["Predicted House Value"]
).abs()

sample_predictions.to_csv(
    "../outputs/sample_predictions.csv",
    index=False
)

print("Sample predictions saved successfully!")
print("File: outputs/sample_predictions.csv")

display(sample_predictions)
```

```python
# Verify saved model predictions

loaded_predictions = loaded_model.predict(X_test)

prediction_difference = np.max(
    np.abs(y_pred_rf - loaded_predictions)
)

print("Maximum prediction difference:", prediction_difference)

if prediction_difference == 0:
    print("Verification successful! Saved model produces identical predictions.")
else:
    print("Verification completed, but predictions differ slightly.")
```
